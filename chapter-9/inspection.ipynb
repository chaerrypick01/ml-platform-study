# PVC 학습 결과 확인
공유 볼륨에서 실제 가중치 평가 결과와 TensorBoard 이벤트를 읽어요.

```python
from pathlib import Path
import json
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator
root = Path('/shared')
results = sorted(root.glob('ch9-*/training/result.json'))
assert results, 'Run the PVC training pipeline first'
result_path = results[-1]
result = json.loads(result_path.read_text())
print('epochs:', result['epochs'])
print('metrics:', result['evaluation']['metrics'])
print('MLflow Run ID:', result['mlflow_run_id'])
events = EventAccumulator(str(result_path.parent / 'tensorboard')).Reload()
assert 'train/box_loss' in events.Tags()['scalars']
print('TensorBoard tags:', events.Tags()['scalars'])
```
